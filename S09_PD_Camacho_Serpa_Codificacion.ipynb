## Por qué usé ordinal encoding en nivel_lealtad

Porque naturalmente esta variable tiene un orden en el contexto real y el de NovaMarket, entre más valioso es el nombre del mineral más valioso es el nivel de lealtad del cliente, si se asumen todos como iguales no tendría sentido llamarlo nivel

## Separación en entrenamiento y prueba

Antes de aplicar el target encoding, dividimos el dataset en dos partes: entrenamiento (80%) y prueba (20%). Esto se hace ahora porque el target encoding necesita calcularse solo con los datos de entrenamiento, si no habría fuga de datos.

```python
from sklearn.model_selection import train_test_split
```

```python
df_entrenamiento, df_prueba = train_test_split(df, test_size=0.2, random_state=42)
```

`monto_compra` tiene algunos valores nulos que arrastramos de semanas anteriores. Los eliminamos solo del entrenamiento porque el `TargetEncoder` no acepta nulos en la variable objetivo.

```python
df_entrenamiento = df_entrenamiento.dropna(subset=["monto_compra"])
```

```python
print("Filas de entrenamiento:", len(df_entrenamiento))
print("Filas de prueba:", len(df_prueba))
```

## Target Encoding de ciudad_tienda

Para `ciudad_tienda` usamos target encoding porque queremos que cada ciudad se represente según el monto de compra promedio de esa ciudad. Así, en vez de tener texto, cada ciudad queda como un número que refleja qué tan altos suelen ser sus montos de compra.

```python
from category_encoders import TargetEncoder
```

```python
codificador_ciudad = TargetEncoder(cols=["ciudad_tienda"], smoothing=10)
codificador_ciudad.fit(df_entrenamiento["ciudad_tienda"], df_entrenamiento["monto_compra"])
```

El `smoothing=10` sirve para que las ciudades con pocos registros no queden con un valor muy raro; las suaviza acercándolas al promedio general.

### Transformar entrenamiento y prueba

Usamos el mismo codificador ya ajustado para transformar los dos conjuntos. No volvemos a hacer `fit()` sobre la prueba, solo `transform()`, así evitamos la fuga de datos.

```python
df_entrenamiento["ciudad_tienda_cod"] = codificador_ciudad.transform(df_entrenamiento["ciudad_tienda"])
df_prueba["ciudad_tienda_cod"] = codificador_ciudad.transform(df_prueba["ciudad_tienda"])
```

```python
print("Entrenamiento:")
print(df_entrenamiento["ciudad_tienda_cod"].describe())
print("Prueba:")
print(df_prueba["ciudad_tienda_cod"].describe())
```

## Agrupamiento de codigo_postal

`codigo_postal` tiene alta cardinalidad (493 valores únicos), entonces no vale la pena aplicarle one-hot ni target encoding directamente. Lo que hacemos es agrupar los códigos poco frecuentes en una categoría llamada 'Otras', quedándonos solo con los códigos que aparecen más de 15 veces en entrenamiento.

```python
conteo_postal = df_entrenamiento["codigo_postal"].value_counts()
```

```python
postales_frecuentes = conteo_postal[conteo_postal >= 15].index
```

Aplicamos la misma lista `postales_frecuentes` (calculada solo con entrenamiento) también en la prueba, para no filtrar información.

```python
df_entrenamiento["codigo_postal_agrupado"] = df_entrenamiento["codigo_postal"].where(df_entrenamiento["codigo_postal"].isin(postales_frecuentes), "Otras")
df_prueba["codigo_postal_agrupado"] = df_prueba["codigo_postal"].where(df_prueba["codigo_postal"].isin(postales_frecuentes), "Otras")
```

```python
print("Códigos únicos originales:", df_entrenamiento["codigo_postal"].nunique())
print("Códigos únicos agrupados:", df_entrenamiento["codigo_postal_agrupado"].nunique())
```
