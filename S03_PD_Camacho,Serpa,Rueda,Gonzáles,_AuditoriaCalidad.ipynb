### **2.2. Exactitud**

En esta sección se audita si los valores presentes en el dataset son **plausibles** desde el punto de vista del negocio. A diferencia de la completitud (que evalúa si el dato existe), la exactitud evalúa si el dato **tiene sentido**: un precio puede estar registrado pero ser negativo, una edad puede aparecer como 200 años, etc.

Para esto se definen **reglas de dominio**: condiciones que un dato debe cumplir para ser aceptado como válido. Cada regla se implementa como un filtro sobre `df` y se cuenta cuántas filas la incumplen con `len()`.

Se aplican **6 reglas**: 4 provenientes del enunciado y 2 propias, propuestas a partir de los hallazgos previos del perfilado.

```python
# Precios negativos
filas = df[df['precio_unitario'] < 0]
print("Precios negativos:", len(filas))
filas
```

#### Precios negativos

Filtré las filas donde *precio_unitario* es menor a 0. Un precio negativo no tiene sentido en un contexto de retail, así que cualquier registro que aparezca aquí es un error de captura.

**Resultado:** 19.629 registros con precio negativo.

```python
# Precios mayores a 10 millones
filas = df[df['precio_unitario'] > 10000000]
print("Precios mayores a 10 millones:", len(filas))
filas
```

#### Precios mayores a $10.000.000

Filtré los registros con *precio_unitario* superior a 10 millones. Aunque NovaMarket vende productos de distintas categorías, un valor unitario tan alto es inusual y sugiere errores de digitación (probablemente ceros de más).

**Resultado:** 6.164 registros con precio absurdamente alto.

```python
# Unidades negativas
filas = df[df['unidades_vendidas'] < 0]
print("Unidades negativas:", len(filas))
filas
```

#### Unidades negativas

Filtré las filas donde *unidades_vendidas* es menor a 0. No es posible vender una cantidad negativa de un producto, así que estos casos también corresponden a errores de captura.

**Resultado:** 18.550 registros con unidades negativas.

```python
# Edad fuera de 0 a 120
filas = df[(df['edad_cliente'] < 0) | (df['edad_cliente'] > 120)]
print("Edades fuera de rango:", len(filas))
filas
```

#### Edades fuera del rango [0, 120]

Filtré los registros con *edad_cliente* menor a 0 o mayor a 120. Usé el operador `|` porque basta con que se cumpla una de las dos condiciones para que la edad sea implausible.

**Resultado:** 27.829 registros con edad fuera de rango, incluyendo valores negativos y edades superiores al límite biológico razonable.

```python
# Unidades mayores a 100
filas = df[df['unidades_vendidas'] > 100]
print("Unidades mayores a 100:", len(filas))
filas
```

#### Unidades mayores a 100 (regla propia)

Además de las reglas del enunciado, propuse esta regla adicional. En un dataset de retail dirigido a consumidor final, un pedido de más de 100 unidades del mismo producto es muy poco frecuente y probablemente indica un error de captura (por ejemplo, valores como 9999 que aparecen como máximo en el describe de *unidades_vendidas*).

**Resultado:** 18.526 registros con más de 100 unidades por pedido.

```python
# Correos sin @
correos_no_nulos = df[df['correo_cliente'].notna()]
filas = correos_no_nulos[~correos_no_nulos['correo_cliente'].str.contains('@')]
print("Correos sin @:", len(filas))
filas
```

#### Correos sin `@` (regla propia)

Como segunda regla propia, revisé el formato de la columna *correo_cliente*. Un correo válido siempre debe contener el símbolo `@`. Hice el filtro en dos pasos:

1. Primero me quedé con los correos que no son nulos, para no mezclar este hallazgo con el de completitud.
2. Sobre esos, filtré los que no contienen `@`.

**Resultado:** 51.532 registros con correo mal formateado (existe la cadena de texto pero no cumple la estructura mínima de un correo electrónico).

#### Hallazgos de exactitud

Se aplicaron 6 reglas de dominio al dataset de NovaMarket. Los resultados fueron:

| # | Regla | Registros afectados |
|---|---|---|
| 1 | *precio_unitario* < 0 | 19.629 |
| 2 | *precio_unitario* > $10.000.000 | 6.164 |
| 3 | *unidades_vendidas* < 0 | 18.550 |
| 4 | *edad_cliente* fuera de [0, 120] | 27.829 |
| 5 | *unidades_vendidas* > 100 (regla propia) | 18.526 |
| 6 | *correo_cliente* sin `@` (regla propia) | 51.532 |

**Total de registros con al menos un problema de exactitud: hasta 142.230**, aunque algunas filas pueden acumular varios errores simultáneamente (por ejemplo, un mismo pedido con precio negativo y edad fuera de rango).

#### Conclusiones

- Las reglas 1, 3 y 4 evidencian **errores de captura básicos**: valores negativos o extremos en campos que nunca deberían tenerlos. En conjunto afectan cerca del 10% del dataset.
- La regla 2 confirma la presencia de **outliers extremos en precio_unitario**, con valores absurdamente altos que deben tratarse antes de cualquier análisis estadístico.
- La regla 5 refuerza el hallazgo del describe de *unidades_vendidas*: valores como 9999 no corresponden a compras reales.
- La regla 6 revela un problema de **integridad estructural** en *correo_cliente* que impacta directamente cualquier campaña de comunicación con clientes: más de 51 mil correos existen pero no son válidos como dirección de correo.
- Combinando completitud y exactitud, se estima que **aproximadamente el 22% del dataset presenta al menos un problema de calidad** (nulos o valores implausibles), lo cual justifica una fase de limpieza sistemática antes de cualquier análisis posterior.

### **2.6. Oportunidad**

La oportunidad evalúa si la información se encuentra disponible y actualizada en el momento adecuado para su uso.

En esta sección se revisan dos aspectos:

- `fecha_pedido`, para identificar registros con fechas futuras que no deberían existir en un dataset operacional.
- `fecha_actualizacion_stock`, para determinar qué tan actualizada está la información de inventario.

Para el stock se establece un umbral de **90 días**: un registro con más de tres meses sin actualización deja de representar bien la disponibilidad actual de un producto. La fecha de corte de la auditoría es el **11 de agosto de 2026**.

```python
# Conversión robusta de fecha_pedido (maneja formatos mezclados)
fecha_txt = df['fecha_pedido'].astype('string').str.strip()

formato_iso = fecha_txt.str.match(r'^\d{4}-\d{1,2}-\d{1,2}$', na=False)
formato_dmy = fecha_txt.str.match(r'^\d{1,2}/\d{1,2}/\d{4}$', na=False)

fecha_convertida = pd.Series(pd.NaT, index=df.index, dtype='datetime64[ns]')
fecha_convertida.loc[formato_iso] = pd.to_datetime(fecha_txt.loc[formato_iso], format='%Y-%m-%d', errors='coerce')
fecha_convertida.loc[formato_dmy] = pd.to_datetime(fecha_txt.loc[formato_dmy], format='%d/%m/%Y', errors='coerce')

# Fecha de referencia de la auditoría
fecha_corte = pd.Timestamp('2026-08-11')

# 1. Pedidos con fecha futura
pedidos_futuros = fecha_convertida.notna() & (fecha_convertida > fecha_corte)
cantidad_pedidos_futuros = pedidos_futuros.sum()
porcentaje_pedidos_futuros = cantidad_pedidos_futuros / len(df) * 100

# 2. Stock desactualizado (> 90 días)
fecha_stock = pd.to_datetime(df['fecha_actualizacion_stock'], errors='coerce')
umbral_dias = 90
fecha_limite = fecha_corte - pd.Timedelta(days=umbral_dias)

stock_desactualizado = fecha_stock.notna() & (fecha_stock < fecha_limite)
cantidad_stock_desactualizado = stock_desactualizado.sum()
porcentaje_stock_desactualizado = cantidad_stock_desactualizado / len(df) * 100

# 3. Stock con fecha futura
stock_futuro = fecha_stock.notna() & (fecha_stock > fecha_corte)
cantidad_stock_futuro = stock_futuro.sum()
porcentaje_stock_futuro = cantidad_stock_futuro / len(df) * 100

resultados_oportunidad = pd.DataFrame({
    'Problema': [
        'Pedidos con fecha futura',
        'Stock con más de 90 días sin actualizar',
        'Actualización de stock futura'
    ],
    'Registros afectados': [
        cantidad_pedidos_futuros,
        cantidad_stock_desactualizado,
        cantidad_stock_futuro
    ],
    'Porcentaje': [
        round(porcentaje_pedidos_futuros, 2),
        round(porcentaje_stock_desactualizado, 2),
        round(porcentaje_stock_futuro, 2)
    ]
})

print("Fecha de corte:", fecha_corte.date())
print("Umbral de oportunidad:", umbral_dias, "días")
print("Fecha límite para stock actualizado:", fecha_limite.date())
print()
print(resultados_oportunidad.to_string(index=False))

print("\nRango de fecha_pedido válida:")
print("Mínima:", fecha_convertida.min())
print("Máxima:", fecha_convertida.max())

print("\nRango de fecha_actualizacion_stock:")
print("Mínima:", fecha_stock.min())
print("Máxima:", fecha_stock.max())
```

##### Hallazgos de oportunidad

Tomando como fecha de corte el **11 de agosto de 2026**, encontramos que en `fecha_pedido` hay **16.485 registros con fechas futuras (2,58 % del dataset)**. Son pedidos que, según la fecha de la auditoría, ni siquiera deberían haber ocurrido — algunos llegan hasta diciembre de 2027.

Para `fecha_actualizacion_stock` fijamos un umbral de 90 días, porque NovaMarket usa la información de inventario para decisiones operativas y un registro con más de tres meses sin actualización deja de reflejar bien la disponibilidad real del producto. Bajo ese criterio, encontramos que **350.611 registros (54,78 %) tienen el stock desactualizado**, lo cual es más de la mitad del dataset y es de lejos el problema más grave de esta dimensión.

Además, aparecen **21.722 registros (3,39 %) con fecha de actualización de stock en el futuro**, lo cual no debería ser posible.

El problema principal está claramente en la actualización del inventario: cualquier decisión de reposición o reporte de disponibilidad construida sobre estos datos va a estar sesgada. Se recomienda establecer una frecuencia máxima de actualización del inventario y controles que impidan registrar fechas posteriores a la fecha real de la operación.

## **4. Priorización**

Cualquier reporte de disponibilidad, decisión de reposición o análisis de rotación construido sobre estos datos va a estar sesgado. Es el problema más masivo del dataset.

**2. Inconsistencias de formato en categóricas (Consistencia).** Casi la mitad de los registros de `ciudad_tienda` (49,20 %) y `categoria_producto` (47,61 %) tienen variantes de escritura de la misma categoría (mayúsculas, tildes, espacios, abreviaturas). Sin fechas confiables no se puede saber cuándo se hizo cada venta; sin correos válidos no llegan facturaciones ni notificaciones al cliente.

**4. Datos faltantes (Completitud).** Faltan el 16,46 % de correos y el 3,38 % de precios/montos. Los precios faltantes rompen los cálculos de ingresos totales; los correos faltantes limitan la comunicación con el cliente.

**5. Valores numéricos imposibles (Exactitud).** Precios negativos, unidades negativas, edades fuera del rango humano y precios de más de 10 millones. Cada uno afecta menos del 5 % individualmente, pero en conjunto distorsionan las agregaciones (medias, sumas, ingresos totales).

**6. Duplicados de pedido (Unicidad).** 20.000 pedidos aparecen dos veces con toda la información idéntica (3,12 %). Genera doble conteo de ventas, unidades e ingresos.

**7. Pedidos con fecha futura (Oportunidad).** 16.485 registros (2,58 %) tienen fechas posteriores a la fecha de corte, algunas hasta 2027. Toca decidir si se corrigen o se eliminan.

## **5. Recomendaciones**

**Fechas.** Convertir `fecha_pedido` a formato estándar (`YYYY-MM-DD`) haciendo un parseo tolerante a los formatos mezclados (`DD/MM/YYYY`, `D/M/YYYY`, `YYYY-MM-DD`). Marcar como nulas las fechas que no se puedan interpretar y revisar caso por caso las que quedan en el futuro. Para `fecha_actualizacion_stock`, establecer un proceso periódico de actualización y un control que impida registrar fechas posteriores a la fecha de operación.

**Correos.** Para los faltantes, asignar un placeholder como `'sin_correo'` o dejarlos nulos según la política del negocio. Para los inválidos (sin `@` o mal formados), enviarlos a un flujo de revisión manual antes de usarlos en campañas de comunicación.

**Numéricos.** Pasar a positivo los valores negativos de `precio_unitario`, `unidades_vendidas` y `edad_cliente` cuando el signo negativo sea claramente un error de captura. Para `precio_unitario` mayor a 10 millones y `unidades_vendidas` mayor a 100, revisar caso por caso — pueden ser errores de digitación (ceros de más) o casos legítimos. Los precios faltantes se pueden imputar por la mediana del producto o categoría.

**Categóricas.** Homologar `ciudad_tienda`, `categoria_producto`, `canal_compra`, `metodo_pago` y `nivel_satisfaccion` quitando espacios al inicio y final, normalizando a minúsculas y luego mapeando cada variante a su categoría estándar. También ajustar el `codigo_postal` según la ciudad correspondiente después de la homologación.

**Unicidad.** Eliminar los 20.000 pedidos duplicados conservando una sola ocurrencia por `id_pedido` y establecer una restricción única en la fuente para que no se puedan volver a registrar.
